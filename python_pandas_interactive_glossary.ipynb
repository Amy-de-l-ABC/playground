## Glossary

Quick starting points: [DataFrame](#scrollTo=term-dataframe) · [shape](#scrollTo=term-shape) · [tuple](#scrollTo=term-tuple) · [assignment & reassignment](#scrollTo=term-assignment) · [reference](#scrollTo=term-reference)

### Library

A **library** is reusable code you can import. **pandas** is a Python library for tabular data; **NumPy** is a library for array operations. `pd` and `np` are convenient aliases, not special keywords.

Related: [DataFrame](#scrollTo=term-dataframe) · [ndarray](#scrollTo=term-ndarray).

```python
import pandas as pd
import numpy as np

print(pd.__name__)  # pandas
print(np.__name__)  # numpy
```

### DataFrame

A **DataFrame** is pandas' two-dimensional table-like object: labeled columns, row index, and possibly different column types. Think **Power Query table** more than plain Excel array.

`sample_df` is our three-row DataFrame. Related: [Series](#scrollTo=term-series) · [shape](#scrollTo=term-shape) · [dtype](#scrollTo=term-dtype).

```python
dataframe = sample_df  # A second name for the SAME DataFrame
print(type(dataframe).__name__)
display(dataframe)
```

### Series

A **Series** is a one-dimensional labeled sequence. Selecting one DataFrame column usually gives a Series.

Related: [DataFrame](#scrollTo=term-dataframe) · [dtype](#scrollTo=term-dtype).

```python
series = sample_df["Chaos_Level"]
print(type(series).__name__)
display(series)

# invalid_series = sample_df["Missing_Column"]  # KeyError: this column doesn't exist
```

### shape

**Shape** is the size of a DataFrame or array, measured along each dimension. For a DataFrame the tuple is **(rows, columns)**. `shape` is an **attribute**: no parentheses needed.

Related: [DataFrame](#scrollTo=term-dataframe) · [tuple](#scrollTo=term-tuple) · [attribute](#scrollTo=term-attribute).

```python
shape = sample_df.shape
print(shape)       # (3, 2)
print(shape[0])    # rows; Python positions start at 0
print(shape[1])    # columns

# shape[0] = 99    # TypeError: tuples don't allow item assignment
```

### tuple

A **tuple** is an ordered collection whose *element positions/references* cannot be changed. `(255, 128, 0)` is one tuple. You can **reassign the variable** to a new tuple; you cannot edit the old tuple's individual elements.

A tuple may contain a mutable object (like a list) that *can* change inside the tuple. Related: [reassignment](#scrollTo=term-assignment) · [mutation](#scrollTo=term-mutation).

```python
rgb = (255, 128, 0)
print(rgb)
rgb = (18, 3, 91)  # Works: the name refers to a NEW tuple
print(rgb)

# rgb[0] = 255     # TypeError: can't replace a tuple element

nested_tuple = ([255], [128], [0])
nested_tuple[0][0] = 18  # Works: changes the list inside the tuple
print(nested_tuple)
# nested_tuple[0] = [18]  # TypeError: can't replace a tuple element
```

### list

A **list** is an ordered, mutable collection. You can replace, add, or remove elements. Square brackets `[]` create a list.

Related: [tuple](#scrollTo=term-tuple) · [mutation](#scrollTo=term-mutation).

```python
snacks = ["bamboo", "biscuits"]
snacks[0] = "tender shoots"  # Change existing item
snacks.append("cake")       # Add an item
print(snacks)

# print(snacks[99])  # IndexError: no item at position 99
```

### dictionary (dict)

A **dictionary** stores key–value pairs. It's the closest Python counterpart to a **JSON object**. Keys let you look up values; they're unique within a dictionary.

Related: [hashable](#scrollTo=term-hashable) · [object](#scrollTo=term-object).

```python
panda_record = {"Name": "Sir Bamboozle III", "Chaos_Level": 10}
print(panda_record["Name"])
panda_record["Chaos_Level"] = 9  # Mutable: update this entry
print(panda_record)

# print(panda_record["Age"])  # KeyError: there is no "Age" key
```

### variable / name

A **name** (often called a variable) refers to a Python object. A name is not a container with the value physically inside it. Think of an Excel defined name, but remember Python does **not** automatically recalculate earlier assignments.

Related: [assignment](#scrollTo=term-assignment) · [reference](#scrollTo=term-reference).

```python
chaos = 10
print(chaos)
chaos = 11  # Now the name refers to a different integer
print(chaos)
```

### assignment / reassignment

**Assignment** associates a name with an object; **reassignment** makes that name refer to something else. In ordinary Python, names can be reassigned even when the object they're referring to is immutable.

Related: [variable](#scrollTo=term-variable) · [tuple](#scrollTo=term-tuple) · [mutation](#scrollTo=term-mutation).

```python
a = [10, 20]     # Assignment
b = a            # A second name for that SAME list
b = [30, 40]     # Reassignment: b now points elsewhere
print("a:", a)
print("b:", b)
```

### reference

A **reference** connects a name (or collection position) to an object. Two names can refer to the *same* mutable list, so a mutation through either name is visible through both. Reassigning one name does **not** redirect the other.

Related: [reassignment](#scrollTo=term-assignment) · [mutation](#scrollTo=term-mutation).

```python
first_list = ["bamboo"]
second_list = first_list
second_list.append("biscuits")  # Mutate the shared list
print(first_list)                # ['bamboo', 'biscuits']

second_list = ["leaves"]        # Reassign only second_list
print(first_list, second_list)
```

### mutation / change in place

**Mutation** means editing an *existing object* rather than making a name point to a different object. Lists and dictionaries are mutable; tuples, strings, and numbers are immutable.

Related: [reference](#scrollTo=term-reference) · [immutable](#scrollTo=term-immutable).

```python
basket = ["bamboo", "leaves"]
original_basket = basket
basket[1] = "biscuits"  # Mutate the SAME list
print(original_basket)  # The change is visible here too

# "bamboo"[0] = "B"  # TypeError: strings are immutable
```

### immutable

An **immutable** object can't have its contents changed in place. Python integers, strings, and tuples are examples. You can still reassign the **name** to a different object.

Related: [tuple](#scrollTo=term-tuple) · [reassignment](#scrollTo=term-assignment).

```python
number = 10
same_number = number
number = 20   # Reassignment, not mutation
print(number, same_number)  # 20 10

word = "panda"
word = "Panda"  # Also reassignment
# word[0] = "P"  # TypeError: string items cannot be changed
```

### object

In Python, an **object** is almost any value: a number, string, list, dictionary, DataFrame, and so on. A JSON *object*, by contrast, corresponds most closely to a Python **dictionary**.

Pandas also has a dtype called `object`; that does **not** necessarily mean JSON objects are in the column. Related: [dictionary](#scrollTo=term-dictionary) · [dtype](#scrollTo=term-dtype).

```python
print(type(42).__name__)              # int
print(type([1, 2]).__name__)          # list
print(type(sample_df).__name__)       # DataFrame
print(type({"panda": 1}).__name__)    # dict
```

### type()

Python's built-in **`type()`** tells you what kind of Python object something is. Compare it with **`dtype`**, which tells you what type of values a NumPy array or pandas Series is designed to hold.

Related: [object](#scrollTo=term-object) · [dtype](#scrollTo=term-dtype).

```python
object_type = type(sample_df)
print(object_type)
print(type(sample_df["Chaos_Level"]).__name__)  # Series
print(sample_df["Chaos_Level"].dtype)            # dtype of its values
```

### dtype

**`dtype`** means **data type**, especially in pandas and NumPy. A DataFrame's columns can have different dtypes. Pandas `object` dtype often means strings or mixed Python objects, not necessarily JSON objects.

Related: [type()](#scrollTo=term-type) · [Series](#scrollTo=term-series) · [ndarray](#scrollTo=term-ndarray).

```python
dtype = sample_df["Chaos_Level"].dtype
print(dtype)           # int64 (typically)
print(sample_df.dtypes)  # dtype for each column

messy_weights = pd.Series([116.4, "~105"])
print(messy_weights.dtype)  # object: numeric + text values mixed
```

### ndarray

A NumPy **`ndarray`** is an **N-dimensional array** (`nd` = *n-dimensional*). It's typically position-based and holds values of one data type. Unlike a DataFrame, it doesn't have pandas-style named columns.

Related: [DataFrame](#scrollTo=term-dataframe) · [shape](#scrollTo=term-shape) · [dtype](#scrollTo=term-dtype).

```python
ndarray = np.array([[1, 2], [3, 4]])
print(ndarray)
print("shape:", ndarray.shape)  # (2, 2)
print("dtype:", ndarray.dtype)

# ndarray[5, 0] = 99  # IndexError: row 5 doesn't exist
```

### hashable

**Hashable** means Python can calculate a stable hash consistent with the object's equality behavior. Dictionary **keys** and set members must be hashable. A tuple of numbers usually is; an ordinary list isn't.

*Analogy:* dictionary keys are **lookup identifiers**, a little like primary keys. But “hashable” is a Python technical property, **not** the database uniqueness constraint itself.

Related: [dictionary](#scrollTo=term-dictionary) · [tuple](#scrollTo=term-tuple).

```python
color_names = {(255, 128, 0): "orange"}
print(color_names[(255, 128, 0)])
print(hash((255, 128, 0)))

# bad_key = {[255, 128, 0]: "orange"}  # TypeError: list is unhashable
# hash(([255], [128], [0]))            # TypeError: tuple contains lists
```

### attribute

An **attribute** is information or functionality accessed using a dot (`.`). `sample_df.shape` gets an attribute; no parentheses. Some attributes refer to methods, which are callable.

Related: [shape](#scrollTo=term-shape) · [method](#scrollTo=term-method).

```python
dimensions = sample_df.shape  # Attribute, not a function call
print(dimensions)

# sample_df.shape()  # TypeError: the returned tuple isn't callable
```

### function / method

A **function** is reusable code you call with parentheses. A **method** is a function accessed through an object (`object.method(...)`). You can use pandas methods without defining classes of your own.

Related: [attribute](#scrollTo=term-attribute) · [DataFrame](#scrollTo=term-dataframe).

```python
row_count = len(sample_df)       # len(...) is a function
preview = sample_df.head(2)      # head(...) is a DataFrame method
print("Rows:", row_count)
display(preview)

# sample_df.head(2, 3)  # TypeError: too many positional arguments
```
